# Day 13: Model Optimization - Overfitting & Regularization

## Objective
* Train a baseline Ordinary Least Squares (OLS) Regression model.
* Apply Ridge (L2) and Lasso (L1) regularization.
* Diagnose overfitting by analyzing the gap between Training and Testing performance.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler

# Set random seed for reproducibility
np.random.seed(42)

# Generate synthetic data: 150 rows, 30 features (but only 10 truly matter)
X, y = make_regression(n_samples=150, n_features=30, n_informative=10, noise=25.0, random_state=42)

print(f"Dataset shape: {X.shape}")
```

```python
# 1. Split into 70% Training and 30% Testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 2. Scale features based on Training data statistics
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Data successfully split and scaled!")
```

```python
# Initialize models with chosen regularization strengths (alpha)
models = {
    "Baseline (OLS)": LinearRegression(),
    "Ridge (L2)": Ridge(alpha=15.0),
    "Lasso (L1)": Lasso(alpha=2.5)
}

results = []

# Train and gather metrics
for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    
    # Predict
    train_pred = model.predict(X_train_scaled)
    test_pred = model.predict(X_test_scaled)
    
    # Calculate metrics
    train_rmse = np.sqrt(mean_squared_error(y_train, train_pred))
    test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))
    train_r2 = r2_score(y_train, train_pred)
    test_r2 = r2_score(y_test, test_pred)
    
    # Count how many features the model actually used (coefficient != 0)
    active_features = np.sum(model.coef_ != 0)
    
    results.append({
        "Model": name,
        "Train RMSE": round(train_rmse, 2),
        "Test RMSE": round(test_rmse, 2),
        "Train R2": round(train_r2, 3),
        "Test R2": round(test_r2, 3),
        "Active Features": f"{active_features}/30"
    })

# Convert summary data to a clean DataFrame
df_results = pd.DataFrame(results)
df_results
```

```python
plt.figure(figsize=(12, 5))

# Plot baseline coefficients
plt.plot(models["Baseline (OLS)"].coef_, label="Baseline (OLS)", marker='o', alpha=0.7)
# Plot Ridge coefficients
plt.plot(models["Ridge (L2)"].coef_, label="Ridge (L2)", marker='s', alpha=0.7)
# Plot Lasso coefficients
plt.plot(models["Lasso (L1)"].coef_, label="Lasso (L1)", marker='^', alpha=0.7)

plt.axhline(0, color='red', linestyle='--', alpha=0.5)
plt.title("How Regularization Affects Model Coefficients")
plt.xlabel("Feature Index")
plt.ylabel("Coefficient Value")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
## Performance Diagnostics

1. **Signs of Overfitting in Baseline (OLS):** 
   * The **Baseline** model shows a massive performance gap. It scores an excellent $R^2$ on training data, but drops significantly on the Test data. This "variance gap" confirms it memorized data noise.
2. **How Ridge Fixed It:** 
   * **Ridge** accepted a slightly worse training score, but its test RMSE dropped sharply compared to OLS. It stabilized the model weights.
3. **How Lasso Fixed It:** 
   * **Lasso** successfully dropped irrelevant features down to zero active features, providing an easily interpretable model that generalizes beautifully.
```
